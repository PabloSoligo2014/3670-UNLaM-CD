```python
import pandas as pd
from functools import reduce
```

```python

# 1. Leer los archivos CSV
df1 = pd.read_csv("./cammesa_historical.csv", parse_dates=["fecha"])
df1['fecha'] = pd.to_datetime(df1['fecha']).dt.tz_localize(None)
df1['fecha_dia'] = df1['fecha'].dt.floor('D') # Normaliza la hora a 00:00:00

```

```python
df2 = pd.read_csv("./cammesa_businessday.csv", parse_dates=["fecha"])
df2['fecha'] = pd.to_datetime(df2['fecha']).dt.tz_localize(None)
df2['fecha_dia'] = df2['fecha'].dt.floor('D')

```

```python
df_merged = pd.merge(
    df1, 
    df2.drop(columns=['fecha']), # Eliminamos 'fecha' de df2 para no duplicar columnas
    on='fecha_dia', 
    how='left' # 'left' preserva la resolución en minutos de df1
)
```

```python
df3 = pd.read_csv("./cammesa_holidays.csv")
df3 = df3.rename(columns={"0": "fecha"})
df_merged['diaFeriado'] = df_merged['fecha_dia'].isin(pd.to_datetime(df3['fecha']).dt.floor('D')).astype(bool)
```

```python
df = df_merged
df.set_index('fecha', inplace=True)
df.to_csv("cammesa_data.csv", index=True)
```
