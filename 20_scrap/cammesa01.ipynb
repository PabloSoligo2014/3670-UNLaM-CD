```python
import pandas as pd
from matplotlib import pyplot as plt
```

```python
df = pd.read_csv("./cammesa_data.csv", parse_dates=["fecha"])
```

```python
df["temp"] = df["temp"].ffill()
df["temp"] = df["temp"].bfill()
```

```python
plt.figure(figsize=(12, 6))
plt.plot(df['fecha'], df['demanda'], label='Demanda', color='blue')
plt.title('Demanda de Energía a lo Largo del Tiempo')   
plt.xlabel('Fecha')
plt.ylabel('Demanda (MW)')
plt.legend()
plt.grid()
plt.show()
```

Correlacion entre temperatura y consumo

```python
plt.figure(figsize=(12, 6))
plt.scatter(df['temp'], df['demanda'], label='Temperatura', color='red')
plt.title('Demanda vs Temperatura')
plt.xlabel('Temperatura (°C)')
plt.ylabel('Demanda (MW)')
plt.legend()
plt.grid()
plt.show()
```
